# Imports & Setup

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.decomposition import PCA
import seaborn as sns
import time
```

Load Dataset from TensorFlow

```python
# Load MNIST directly from tensorflow.keras.datasets
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Train shape:", x_train.shape, y_train.shape)
print("Test shape:", x_test.shape, y_test.shape)
```

Visualize Some Samples

```python
plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap='gray')
    plt.title(f"Label: {y_train[i]}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```

# Preprocess Data

```python

x_train_flat = x_train.reshape(x_train.shape[0], -1)
x_test_flat = x_test.reshape(x_test.shape[0], -1)


x_train_flat = x_train_flat.astype('float32') / 255.0
x_test_flat = x_test_flat.astype('float32') / 255.0


N_TRAIN = 10000
N_TEST = 2000

x_train_small = x_train_flat[:N_TRAIN]
y_train_small = y_train[:N_TRAIN]
x_test_small = x_test_flat[:N_TEST]
y_test_small = y_test[:N_TEST]

print("Using train:", x_train_small.shape, " test:", x_test_small.shape)
```

# (Optional / recommended) Speed up with PCA

```python

pca = PCA(n_components=100, random_state=42)
x_train_pca = pca.fit_transform(x_train_small)
x_test_pca = pca.transform(x_test_small)

print("Explained variance retained:", sum(pca.explained_variance_ratio_))
```

# Train KNN Classifier

```python
k = 5
knn = KNeighborsClassifier(n_neighbors=k, n_jobs=-1)

start = time.time()
knn.fit(x_train_pca, y_train_small)
print(f"Training took {time.time() - start:.2f} seconds")
```

# Evaluate

```python
start = time.time()
y_pred = knn.predict(x_test_pca)
print(f"Prediction took {time.time() - start:.2f} seconds")

acc = accuracy_score(y_test_small, y_pred)
print(f"Accuracy: {acc * 100:.2f}%")

print("\nClassification Report:")
print(classification_report(y_test_small, y_pred))
```

# Confusion Matrix

```python
cm = confusion_matrix(y_test_small, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```

# Test on Custom Samples & Visualize Predictions

```python
plt.figure(figsize=(12, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test_small[i].reshape(28, 28), cmap='gray')
    color = 'green' if y_pred[i] == y_test_small[i] else 'red'
    plt.title(f"Pred: {y_pred[i]} | True: {y_test_small[i]}", color=color)
    plt.axis('off')
plt.tight_layout()
plt.show()
```

# Find Best K Value

```python
k_values = [1, 3, 5, 7, 9, 11]
accuracies = []

for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k, n_jobs=-1)
    model.fit(x_train_pca, y_train_small)
    pred = model.predict(x_test_pca)
    acc = accuracy_score(y_test_small, pred)
    accuracies.append(acc)
    print(f"k={k}: accuracy={acc*100:.2f}%")

plt.plot(k_values, accuracies, marker='o')
plt.xlabel('k')
plt.ylabel('Accuracy')
plt.title('KNN Accuracy vs k')
plt.grid(True)
plt.show()
```

# Visualize How KNN Actually Decides (Nearest Neighbors Explained)

```python

sample_idx = 0
sample_image = x_test_pca[sample_idx].reshape(1, -1)

distances, indices = knn.kneighbors(sample_image, n_neighbors=5)

plt.figure(figsize=(12, 3))

plt.subplot(1, 6, 1)
plt.imshow(x_test_small[sample_idx].reshape(28, 28), cmap='gray')
plt.title(f"Query\nTrue: {y_test_small[sample_idx]}")
plt.axis('off')

for i, idx in enumerate(indices[0]):
    plt.subplot(1, 6, i + 2)
    plt.imshow(x_train_small[idx].reshape(28, 28), cmap='gray')
    plt.title(f"Neighbor {i+1}\nLabel: {y_train_small[idx]}\nDist: {distances[0][i]:.2f}")
    plt.axis('off')

plt.tight_layout()
plt.show()
print("This is literally how KNN decides: it looks at the 'k' closest training")
print("images (by distance) and picks the most common label among them.")
```

## Self Check — Per-Class Performance
Overall accuracy (93%) doesn't tell us everything — some digits are easier for KNN to recognize than others.
Ei part e amra dekhbo kon digit shobcheye kom accurate predict hocche, ebong keno.

```python

per_class_acc = cm.diagonal() / cm.sum(axis=1)

for digit in range(10):
    print(f"Digit {digit}: {per_class_acc[digit]*100:.2f}% accuracy")

worst_digit = per_class_acc.argmin()
best_digit = per_class_acc.argmax()
print(f"\nWeakest digit: {worst_digit} ({per_class_acc[worst_digit]*100:.2f}%)")
print(f"Strongest digit: {best_digit} ({per_class_acc[best_digit]*100:.2f}%)")
```

```python

confused_with = cm[worst_digit].copy()
confused_with[worst_digit] = 0  
most_confused_digit = confused_with.argmax()

print(f"Digit {worst_digit} shobcheye beshi confuse hoy Digit {most_confused_digit} er sathe")
print(f"({confused_with[most_confused_digit]} ta test sample e eita hoyeche)")


examples_worst = np.where(y_test_small == worst_digit)[0][:5]
plt.figure(figsize=(10, 2))
for i, idx in enumerate(examples_worst):
    plt.subplot(1, 5, i+1)
    plt.imshow(x_test_small[idx].reshape(28,28), cmap='gray')
    color = 'green' if y_pred[idx]==worst_digit else 'red'
    plt.title(f"Pred:{y_pred[idx]}", color=color)
    plt.axis('off')
plt.suptitle(f"Digit {worst_digit} er kichu test example")
plt.tight_layout()
plt.show()
```

### Self Check Questions
Nijeke ei question gulo jiggesh koro, tarpor answer gulo dekho:

**Q1: Keno kichu digit (jemon 4, 9, othoba 3, 8) onek shomoy confuse hoy?**
Answer: Karon egulo dekhte structurally kachakachi — jemon 4 ar 9 er upper loop, othoba 3 ar 8 er curve gulo
similar shape er, tai pixel-distance (jeta KNN use kore) e egulo kacha kachi pore jay.

**Q2: PCA use korar por accuracy kome jayni keno?**
Answer: Karon 784 dimension er modhye onek dimension actually kono kaje lage na (empty/border pixel gulo shob
image e shomoi kalo thake) — PCA shudhu important variation gulo rekhe dey, tai accuracy pray same thake
kintu speed onek beshi hoy.

**Q3: k baranor por accuracy kome gelo keno (k=1 e 93.5%, k=11 e 92.3%)?**
Answer: Choto k (jemon k=1) hocche "overfitting" er kach kachi — shudhu 1ta nearest neighbor er upor
depend kore. Boro k (jemon k=11) onek beshi neighbor er "vote" nay, tai kichu kichu somoy alada digit
er neighbor o vote e chole ashe, accuracy shamanno kome jay. Ei dataset er jonno k=1 ba k=3 e best result.

## Conclusion

**Ki kora holo:** MNIST handwritten digit dataset (60,000 training + 10,000 test image, 28x28 grayscale)
theke ekটA subset (10,000 train / 2,000 test) niye KNN (K-Nearest Neighbors) classifier train kora hoyeche.

**Result:**
- PCA diye 784 dimension theke 100 dimension e reduce kora hoyeche (speed onek beshi baseche, accuracy pray
  same theke geche)
- Best result peyechi **k=1 e 93.50% accuracy**, jodio k=3 ebong k=5 o kach kachi (93.10%) result diyeche
- Sob theke shohoje distinguish kora jay emon digit ebong sob theke confuse hoy emon digit gulo alada kore
  dekhano hoyeche (per-class accuracy diye)
- Normalization (pixel value 0-255 theke 0-1 e scale kora) er effect o দেখানো হয়েছে

**Limitation / Future Improvement:**
- Shudhu 10,000/2,000 sample byabohar kora hoyeche speed er jonno — full dataset (60,000/10,000) byabohar
  korle accuracy aro beshi hote pare (kintu KNN onek slow hoye jabe boro dataset e)
- CNN (Convolutional Neural Network) ei dataset e onek beshi accuracy dey (~99%), karon eta image er
  spatial structure bujhte pare, jekhane KNN shudhu raw pixel distance dekhe
- Distance metric change kore (jemon 'manhattan' ba 'cosine') accuracy compare kora jete pare
